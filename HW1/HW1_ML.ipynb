# Augmenting MNIST images with noisy pixels

```python
import random
mnist['data'] = mnist['data'].astype(int)
# creating a list that contains augmented images corresponding to the training data
train = []

# Augmenting each of the training images in mnist dataset with 280 noisy pixels
for i in range(len(mnist['data'])):
  pixels = []
  for j in range(280):
    #intensity of random pixel varies between 0 and 255
    pixels.append(random.randint(0,255))
  train.append(np.hstack((mnist['data'][i], pixels)))

train = np.array(train)
for i in range(len(train)):
  train[i] = np.array(train[i])


mnist['testdata'] = mnist['testdata'].astype(int)

# creating a list that contains augmented images corresponding to the testing data
test = []

for i in range(len(mnist['testdata'])):
  pixels = []
  for j in range(280):
    pixels.append(random.randint(0,255))
  test.append(np.hstack((mnist['testdata'][i], pixels)))

test = np.array(test)
for i in range(len(test)):
  test[i] = np.array(test[i])


```

# Deploying Nearest Neighbour model for classification

```python
from sklearn.neighbors import KNeighborsClassifier
# Deploying the Nearest Neighbor classifier
model = KNeighborsClassifier(1)
# Training the Nearest Neighbor classifier over training data
model.fit(train, mnist['labels'])
```

```python
# calculating accuracy of predictions using the testing data
accuracy = model.score(test,mnist['testlabels'])
```

```python
print("Test Error Rate: ", round((1-accuracy)*100, 2), "%")
```

# Obtaining Density vs Sepal Length plots

```python
import math
# matplotlib used for creating the gaussian plots
import matplotlib.pyplot as plt
#print(iris['data'])
min = 1000000
for i in range(len(iris['data'])):
  if(iris['data'][i][0] < min):
    min = iris['data'][i][0]

# min - the minimum sepal length in iris dataset
print("Minimum Sepal Length in Iris dataset: ", min)

max = 0
for i in range(len(iris['data'])):
  if(iris['data'][i][0] > max):
    max = iris['data'][i][0]

#max - maximum sepal length in iris dataset
print("Maximum Sepal Length in the Iris dataset: ", max)

min = min.astype('float')
max = max.astype('float')
# set a range of 'x' for the generative modelling
# the sepal length is varied between the minimum and the maximum
x = [min, max]
ycor1 = []
xcor = []
y1 = []
while(min<=max):
  xcor.append(min)
  # creating the classifier using parameters obtained through MLE, provided in the question
  f = (1/3)*(1/math.sqrt(2*math.pi*0.68*0.68))*math.exp(-1*pow((min - 6.61),2)/(2*0.68*0.68))
  ycor1.append(f)
  min+=0.01
  y1.append((min, f))

#plt.plot(xcor, ycor1, label = 'density vs sepal length plot', color = 'blue')
```

```python
min = x[0]
max = x[1]
xcor = []
ycor2 = []
y2 = []
while(min<=max):
  xcor.append(min)
  # creating the classifier using parameters obtained through MLE, provided in the question
  f = (1/3)*(1/math.sqrt(2*math.pi*0.47*0.47))*math.exp(-1*pow((min - 5.93),2)/(2*0.47*0.47))
  ycor2.append(f)
  min+=0.01
  y2.append((min, f))

plt.plot(xcor, ycor1, label = 'density vs sepal length plot', color = 'blue')
plt.plot(xcor, ycor2, label = 'density vs sepal length plot', color = 'red')
plt.xlabel("Sepal Length")
plt.ylabel("Density")
plt.title("Density vs Sepal Length plot for the different species of iris")
intersection = 0
#finding the point of intersection between f = 2 and f = 3
min = 100000000

for x in range(len(y1)):
    if(abs(y1[x][1] - y2[x][1]) < min and y1[x][0] >= 5 and y1[x][0] <= 7):
      intersection = y1[x][0]
      min = abs(y1[x][1] - y2[x][1])


print("Intersection point between f = 2 and f = 3: ", intersection)
print("Therefore, a = ", intersection)
print("b = ", max)
```
