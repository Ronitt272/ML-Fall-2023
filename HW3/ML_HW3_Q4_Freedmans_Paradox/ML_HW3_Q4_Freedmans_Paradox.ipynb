```python
import pickle
freedman = pickle.load(open('freedman.pkl', 'rb'))
```

```python
print(freedman['labels'])
```

```python
print(freedman['data'])
print(len(freedman['data'][0]))
print(len(freedman['labels']))
```

```python
p_hat = []

for j in range(1000):
  sum = 0.0
  for i in range(100):
    sum = sum + freedman['data'][i][j]*freedman['labels'][i]
  p_hat_j = sum/100
  p_hat.append(p_hat_j)

print(p_hat)
print(len(p_hat))
```

#Part-A

```python
n = 100
correlation_threshold = 0.2
j_hat = []
for j in range(len(p_hat)):
  p_hat_j = p_hat[j]
  if abs(p_hat_j) > correlation_threshold:
    j_hat.append(j)

print(j_hat)
```

```python
# Part-A answer
print("number of features in j_hat after step-2: ", len(j_hat))
```

#Part-B

```python
train_data = []
for i in range(100):
  row = []
  for j in j_hat:
    row.append(freedman['data'][i][j])
  train_data.append(row)

print(train_data)
print(len(train_data))
print(len(train_data[0]))
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.array(train_data)
y = np.array(freedman['labels'])
reg = LinearRegression()
reg.fit(X,y)
print(reg.coef_)
print(len(reg.coef_))
print(reg.intercept_)
```

```python
risk_w_hat = 0.0
SSE = 0.0
import math
for i in range(100):
  sse = 0.0
  for j in range(42):
    sse = sse + (reg.coef_[j]*train_data[i][j])
  sse = sse - freedman['labels'][i]
  #print(sse)
  SSE = SSE + pow(sse,2)

risk_w_hat = SSE/100
risk_w_hat = round(risk_w_hat,3)
print("The emperical risk of w_hat after step 3: ", risk_w_hat)
print("We see that this value is much smaller than 1, which is the closeby estimate of the variance of the labels on this dataset")
```
